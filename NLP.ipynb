```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
#download the necessry resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
#define a sample text
text = "THE CAT SAT ON THE MAT. THE DOG LAY ON THE RUG . THE CAT CHASED THE DOG"
print("Original Text :",text)
```

```python
#tokenization(splitting the text into words)
tokens =  word_tokenize(text)
print("Tokenized text: ",tokens)
```

```python
#stopword removal (removing commom words like the is was etc)
sample_stop_words = set(stopwords.words("english"))
stopword_removed_words = [eachword for eachword in tokens
if eachword.lower()
not in sample_stop_words]
print('Stopword removed words :',stopword_removed_words)
```

```python
#stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(eachword) for eachword in stopword_removed_words]
print(stemmed_words)
```
