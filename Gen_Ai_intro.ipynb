Trying a simple implementation of an autorecursive gpt2 model

```python
#import pipeline fn from transformer librray
from transformers import pipeline
```

```python
#createa a text gen pipeline using "gpt2" nodel
generator = pipeline("text-generation",model="gpt2")
```

```python
print(generator("someone is trying to kill me",max_length=100))
```

```python
print(generator("tell a story",max_length=100))
```

```python
print(generator("twinkle twinkle little star",max_length=100))
```
