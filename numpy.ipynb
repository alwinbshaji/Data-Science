```python
#create  sample numpy arrays of different dimensions
#for that ,definr a range with a constant seed value
range=np.random.default_rng(seed=500)
#create 1d,2d,3d np arrays
#1D-with 1 row and 5 cols
my1dnparray=range.integers(10,size=5)
#2D-with 5 rows and 4 colms
my2dnparray=range.integers(10,size=(5,4))
#3D-with 5 rows and 4 colms with 3 dims
my3dnparray=range.integers(10,size=(3,5,4))
print(my1dnparray)
print(my2dnparray)
print(my3dnparray)
```

```python
#array slicing in numpy
my1dnparray=np.array([2,3,4,5,6,7,8,8,9,6])
print(my1dnparray[3:]) #all elements from 3rd index or 4th number
print(my1dnparray[:3]) #from initial to the 2nd index or 3rd umber
print(my1dnparray[2:6])
print(my1dnparray[0:9:2])
print(my1dnparray[::2])
print(my1dnparray[::-1])
```

```python
import numpy as np

# Create a 1D NumPy array
my1dnparray = np.array([1, 2, 3, 4, 5, 6, 7])

# Check original array
print("Original Array:")
print(my1dnparray)
print("Original Shape:", my1dnparray.shape)


# Convert 1D array into 2D array
# Method 1: Using reshape()

mynew1darray = my1dnparray.reshape(7, 1)

print("\nMethod 1 - Using reshape():")
print(mynew1darray)
print("Shape:", mynew1darray.shape)


# Method 2: Using np.newaxis

mynew1darray = my1dnparray[:, np.newaxis]

print("\nMethod 2 - Using np.newaxis:")
print(mynew1darray)
print("Shape:", mynew1darray.shape)
```

Array Concatenation

```python
myarr1= np.array([1,2,3])
myarr2= np.array([4,5,6])
myarr3= np.array([7,8,9])

#using concatenation() fn giev array as a list[]
print(np.concatenate([myarr1,myarr2,myarr3]))

#concatenate along the two axis
#concatente along first axis(y)

print(np.concatenate([myarr1,myarr1]))
print(np.concatenate([[myarr1],[myarr2]]))
print(np.concatenate([[myarr1],[myarr2]],axis=1))
```

```python
myarray=[1,2,3,4,5,6,7,8,9]
#split the array intp 3 parts
#split the first part at 2nd index
#split 5th and till  end
split1,split2,split3=np.split(myarray,[2,5,])
print(split1,split2,split3)

#Create a 2d array using the range function
myarray=np.arange(25).reshape((5,5))
print(myarray)

#verTICAL SPLIT
toparray,bottomarray= np.vsplit(myarray,[2])
print(toparray,bottomarray)

#HOORIZONTAL SPLIT
leftarray,rightarray= np.hsplit(myarray,[2])
print(leftarray,rightarray)
```

universal functions

```python
#used to perform vertorized operations (direct operations)
#on an array
#its fast and efficient compared to the typical python list operations

#Create an array of 5 items
array1=np.arange(5)
print(array1)

#perform additon on the umpy array
#add a scalar single value to the array

array2= array1+5
print(array2)

#add two or more arrays together
array3= array1 + array2
print(array3)

#another option
array3 = np.add(array1,array2)
print(array3)

#similarly we can use subtract(),multiple(),divide()
#negative(),floor_division(),power(),mod() math functions

#we can also use trignometric functions , log functions
#exp functions

angles= np.array([45,80,120,90])
print("sin of angles =", np.sin(angles))
```

```python
#reduce and accumulate functions in numpy

myarray = np.arange(10)
print(myarray)

print("reduce array: ", np.add.reduce(myarray))
print("Accumulated array: ", np.add.accumulate(myarray))

#we can also use subtract , multiply,divide etc
```

```python
#aggreagate function in numpy

myarray = np.arange(1,10)
print(myarray)

print("sum :",np.sum(myarray))
print("sum :",np.nansum(myarray))
# nansafe - will not error if any balue is nan(not a number or blank)

print("product :",np.prod(myarray))
print("product :",np.nanprod(myarray))


print("Mean :",np.mean(myarray))
print("Mean :",np.nanmean(myarray))

print("STD Deviation :",np.std(myarray))
print("STD Deviation  :",np.nanstd(myarray))

print("variance :",np.var(myarray))
print("variance  :",np.nanvar(myarray))

print("min :",np.min(myarray))
print("max :",np.max(myarray))
print("index of min value :",np.argmin(myarray))
print("median  :",np.median(myarray))
print("Percentile :",np.percentile(myarray,75))

#for boolean array,will be true if any of the value is true
print("if all elements are true  :",np.any(myarray))

#for boolean array,array will be true if all the values are true
print("if all elements are true  :",np.all(myarray))
```

```python
import pandas as pd
import numpy as np

# Step 1: Create student data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve',
             'Frank', 'Grace', 'Helen', 'Ian', 'Jack'],

    'Height(cm)': [169, 156, 184, 172, 145,
                   183, 163, 187, 175, 171]
}

# Step 2: Convert data into DataFrame
mydata1 = pd.DataFrame(data)

# Step 3: Save data as CSV file
mydata1.to_csv('stud.csv', index=False)

# Step 4: Read CSV file
mydata1 = pd.read_csv('stud.csv')

print("Student Details:")
print(mydata1)

# Step 5: Convert Height column into NumPy array
stud_data = np.array(mydata1['Height(cm)'])

# Step 6: Calculate aggregation functions

print("\nAverage:", np.average(stud_data))

print("Mean:", np.mean(stud_data))

print("Standard Deviation:", np.std(stud_data))

print("25th Percentile:", np.percentile(stud_data, 25))

print("50th Percentile (Median):", np.percentile(stud_data, 50))

print("75th Percentile:", np.percentile(stud_data, 75))
```
