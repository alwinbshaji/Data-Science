matplotlib - the most popular python data visualization library

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#create a sample dataframe
popularity = {"character":["spiderman","spiderman",
                           "superman","superman",
                           "ironman","ironman","ironman",
                           "jerry","jerry","tom","tom"],
              "popularity":["1st","2nd","3rd","4th",
                            "5th","5th","1st","2nd",
                            "3rd","4th","5th"],
              "score1":[60,57,46,59,44,45,60,43,43,59,47],
              "score2":[60,57,46,59,45,47,60,57,42,59,47],
              "gender":["male","male","male","male","male",
                        "male","male","female","female","male",
                        "male"]}

df1= pd.DataFrame(popularity)
df1
```

```python
# a line plot - its the dafault plot

plot1= df1.plot("character")
```

```python
# a simple histogram - will represent char vs scaled version
#of numerical values
hist1 = df1["character"].hist()
```

```python
#for groupeed plots , we have to manually group the data
# and then supplu it to the plot

grouped = df1.groupby("character").sum(numeric_only=True)
```

```python
grouped
```

```python
grouped.plot()
```

```python
#plotting a bar plot
grouped.plot(kind='bar')
plt.title("bar plot of character vs score")
plt.xlabel("Characters")
plt.ylabel("Scores")
plt.show()
```

```python
#pie char
plt.pie(grouped['score1'],labels=grouped.index,autopct='%1.1f%%')

plt.title("Pie chart of scores (score 1)")
plt
```
