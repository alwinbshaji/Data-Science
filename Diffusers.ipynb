```python
import torch
from diffusers import DiffusionPipeline
```

```python
#choose the device id gpu available use it else use cpu
device ='cuda' if torch.cuda.is_available() else "cpu"
```

```python
#iF gpu them data type shoul br float 16 else float 32
dtype = torch.float16 if device =="cuda" else torch.float32
```

```python
# we will be using the opensourced stable diffusion 1.5 model
# its pretrained and available to download from huggingface
pipeline = DiffusionPipeline.from_pretrained("stable-diffusion-v1-5/stable-diffusion-v1-5",
                                             torch_dtype=dtype)
pipeline = pipeline.to(device)
```

```python
#now we have the pretrined model download from hugging face
#we can genereate images using this model
prompt =" generate a realistic picture of three cats playing with football "
image = pipeline(prompt).images[0]
image
```
