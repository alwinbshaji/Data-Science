# **Time Series Analysis**

```python
#time series in padnas can be represented as a data series
#create a data series with hourly datas starting frin 10/5/2026 to
sales_ds = pd.date_range(start='10/05/2026',end='10/15/2026',freq='H')
print(sales_ds)
print(len(sales_ds))
```

```python
#to further process the data converrt the tim eseries into dataframe
#create a dataframe with column name 'date and save this data
sales_df = pd.DataFrame(sales_ds,columns=['date'])
sales_df.head(2)
```

```python
import numpy as np
import pandas as pd
```

```python
#now create another cokumn called 'sales' in our sales_df
#fill that column with random values from 0 to 500
sales_df['sales'] = np.random.randint(0,500, size=(len(sales_df)))
```

```python
sales_df.head(2)
```

```python
#just like how we set primary key column for teable
#here we have to dt the date column as index column
sales_df = sales_df.set_index('date')
```

```python
#to get specifix index values using loc[]
print(sales_df.loc['2026-10-05 01:00:00','sales'])
print(sales_df.loc['2026-10-05 01:00:00'])
```

```python
#get the sales data between two dates
print(sales_df.loc['2026-10-15':'2026-10-06'])
```

```python
#time series resampling
#using statistical functions over different time intervals
#stats function(summary states) :mean(),sum(),count(),min(),max()
#time intr(resampling ,freq):'T','H','D','B','W','M','Q','A'
#Downsampling - reducing to lower frequency time interval
#upsmapling - increasing to higher freq time interval
```

```python
#downsampling perhour slaes data to mean of saled daily
sales_df.resample('D').mean()
```

```python
print(salesdf.isnull().sum())
```

```python
salesdf = salesdf.dropna()
salesdf
```

```python
print(" total revenue = ",salesdf['Revenue'].sum())
```

```python
salesdf['date'] = pd.to_datetime(salesdf['date'])

salesdf['Month'] = salesdf['date'].dt.to_period('M')

monthly_sales = salesdf.groupby('Month')['Revenue'].sum()

average_monthly_sales = monthly_sales.mean()

print("\nAverage Sales Revenue Per Month:")
print(average_monthly_sales)
```

```python
product_sales = salesdf.groupby('Product')['Units Sold'].sum()

best_product = product_sales.idxmax()
best_units = product_sales.max()

print("\nBest Selling Product:", best_product)
print("Number of Units Sold:", best_units)
```

```python
highest_month = monthly_sales.idxmax()
highest_revenue = monthly_sales.max()

print("\nMonth With Highest Sales:", highest_month)
print("Highest Monthly Revenue:", highest_revenue)
```
